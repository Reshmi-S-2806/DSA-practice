```python
# TWO SUM

# create a dictionary
# check for what number is need to acheive a target
# then check if it is in dictionary
# if yes return that,else add into dict

l=[2,7,8,9]
t=9
def twosum(l,t):
  seen={}
  for i,num in enumerate(l):
    complement=t-num
    if complement in seen:
      return [seen[complement],i]
    seen[num]=i
  return []
print(twosum(l,t))
```

```python
# check palindrome

# assign left and right
# check if left and right matched if yes return true else return false
# left +=1,right-=1

def palindrome(str):
  left=0
  right=len(str)-1
  while left<right:
    if str[left]!=str[right]:
      return False
    left+=1
    right-=1
  return True
print(palindrome("racecar"))
print(palindrome("hello"))
print(palindrome("madam"))
```

```python
# valid anagram

# Logic:
# 1. Count each character in the first string.
# 2. Decrease the count for each character in the second string.
# 3. If a character is missing or its count becomes negative, return False.

def anagram(s,t):
  if len(s)!=len(t):
    return False
  freq={}
  for i in s:
      freq[i]=freq.get(i,0)+1
  for i in t:
    if i not in freq:
      return False

    freq[i]-=1
    if freq[i]<0:
      return False
    return True
print(anagram("anagram","nagaram"))
print(anagram("rat","car"))
```
