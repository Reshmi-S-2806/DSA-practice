```python
#two sum

l=[2,7,11,2]
t=9
def twosum(l,t):
  seen={}
  for i,num in enumerate(l):
    complement=t-num
    if complement in seen:
      return [seen[complement],i]
    seen[num]=i
  return []
print(twosum(l,t))
```

```python
#check palindrome

def palindrome(str):
  left=0
  right=len(str)-1
  while left<right:
    if str[left]!=str[right]:
      return False
    left+=1
    right-=1
  return True
print(palindrome("racecar"))
print(palindrome("hello"))
print(palindrome("madam"))
```
